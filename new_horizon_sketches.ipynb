## Key Component Summary

| Component | Function | Details |
|-----------|----------|----------|
| **Main Bus** | Structure & avionics | Hexagonal aluminum frame houses all systems |
| **High-Gain Antenna (HGA)** | Communications | 2.1 m diameter, sends/receives data with Earth |
| **Solar Panels (2)** | Power generation | Provides electrical power during Kuiper Belt operations |
| **RTG** | Power generation | Radioisotope Thermoelectric Generator (~180 W at launch) |
| **Main Engine** | Propulsion | Hydrazine-fueled engine for trajectory corrections |
| **RCS Thrusters (4)** | Attitude control | Reaction Control System for spacecraft orientation |
| **LORRI** | Imaging | Long Range Reconnaissance Imager (telescope, 20 cm aperture) |
| **Ralph** | Multispectral imaging | Visible and infrared imaging |
| **Alice** | Spectroscopy | Ultraviolet imaging spectrometer |
| **SWAP** | Plasma measurement | Solar Wind Around Pluto particle detector |
| **Instrument Boom** | Support | ~4.8 m boom keeps instruments away from spacecraft for thermal reasons |
| **Star Trackers (2)** | Navigation | Use stars for precise spacecraft orientation |
| **Reaction Wheels (3)** | Attitude control | Help orient spacecraft without fuel |
| **Student Dust Counter** | Science | Measures dust impacts (student-built instrument) |
