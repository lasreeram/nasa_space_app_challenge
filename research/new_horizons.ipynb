## Mission Overview

New Horizons was the first spacecraft to explore Pluto and its moons up close (flyby on July 14, 2015). In early 2019 it flew past Arrokoth (2014 MU69), the most distant object ever explored up close. It was the first mission in NASA's New Frontiers program (which also includes Juno and OSIRIS-REx).

| Item | Detail |
|---|---|
| Objectives | Pluto flyby, Kuiper Belt Object flyby |
| Management | NASA / Johns Hopkins University Applied Physics Laboratory (APL) |
| Spacecraft mass | 1,054 lb (478 kg) |
| Power | RTG, ~250 W at launch, ~200 W at Pluto (a Cassini spare) |
| Launch vehicle | Atlas V 551 (AV-010), Cape Canaveral LC-41 |
| Launch | Jan. 19, 2006, 19:00:00 UT |
| Launch velocity | ~36,400 mph (58,536 km/h), highest launch speed of any human-made object relative to Earth |

### Key dates
- **Jan. 19, 2006**: Launch
- **Feb. 28, 2007**: Jupiter gravity assist (+~9,000 mph, shortened the trip by ~3 years)
- **July 14, 2015**: Pluto flyby
- **Jan. 1, 2019**: Arrokoth flyby
- **Oct. 1, 2024**: 60x Earth's distance from the Sun (twice as far as Pluto at flyby)

### Scientific instruments
**Remote-sensing instruments**
- **Ralph**: main color imager and infrared mapping spectrometer. Maps surface composition (nitrogen, methane and water ices) and takes color images. Two parts: MVIC (visible color camera) and LEISA (near-infrared spectral imager).
- **Alice**: ultraviolet imaging spectrometer. Studies the composition and structure of Pluto's atmosphere, looks for escaping gas, and searches for atmospheres around other objects.
- **LORRI** (Long-Range Reconnaissance Imager): telescopic black-and-white camera with a ~20 cm mirror. Takes the highest-resolution images, spots targets from far away, and helps with navigation.

**Radio science**
- **REX** (Radio Science Experiment): uses the spacecraft's radio link to Earth instead of a separate sensor. Measures atmospheric temperature and pressure and surface temperature, and helps determine mass.

**Particle instruments**
- **SWAP** (Solar Wind Around Pluto): measures how the solar wind interacts with Pluto's atmosphere, including how fast the atmosphere escapes.
- **PEPSSI** (Pluto Energetic Particle Spectrometer Science Investigation): detects energetic ions, especially from Pluto's atmosphere, and measures their energy and composition.

**Dust**
- **SDC** (Student Dust Counter): student-built and student-operated dust detector. Counts dust grains along the whole trajectory, mapping dust density from the inner solar system into the Kuiper Belt.

## Flyby Timing: the Timing Knowledge Update (TKU) and Navigation

*Source note: the TKU description comes from the researcher's own notes. The navigation details are background knowledge and not verified against NASA pages. The NSSDCA confirms only the DSN X-band link and the onboard star cameras.*

### The problem
New Horizons flew by Pluto on a pre-loaded command sequence (the "core load") timed to its onboard clock. The cameras and spectrometers are fixed to the spacecraft body, so they point wherever the whole spacecraft points. To center Pluto and its moons in the instrument boresight, the spacecraft had to reach closest approach within +/-540 s (+/-9 min) of the planned time. Radio signals took about 4.5 hours each way, so live control was impossible.

### The Timing Knowledge Update
- The TKU adjusts the onboard clock so the spacecraft believes it is slightly ahead of or behind its real position in the core load.
- All planned flyby activities slide earlier or later together, by up to 540 s, to line up with the final, best predicted arrival time.
- It changes one number instead of rewriting a tested sequence, so activities keep the same order and spacing.

### Navigation data
1. **Radio tracking:** Deep Space Network antennas measure Doppler shift (speed along the line of sight) and ranging (distance), fixing the trajectory relative to Earth.
2. **Optical navigation:** in the months before the encounter, the spacecraft imaged Pluto and its moons against known star fields with LORRI (see the LORRI section below), giving the direction to Pluto with more precision than Earth-based tracking.
3. **Combination:** the navigation team combined both into a best predicted arrival time, which set the TKU and any course corrections.

```mermaid
flowchart LR
    A[DSN Doppler and ranging] --> C[Orbit determination]
    B[Images of Pluto against star background] --> C
    C --> D[Best predicted arrival time]
    D --> E[TKU: shift onboard clock up to 540 s]
    C --> F[Course corrections, if needed]
    E --> G[Sequence aligned with true arrival: Pluto and moons in boresight]
```

### Facts from NASA's NSSDCA (2006-001A)
- Closest approach: 11:48:45 UT on July 14, 2015, at 12,500 km from Pluto. The NSSDCA gives 11 km/s relative velocity, but Wikipedia gives 13.78 km/s (49,600 km/h) at closest approach. The sources disagree, so treat the exact speed as unresolved until checked against an APL or NASA mission document.
- Closest approach to Charon: about 27,000 km.
- Round-trip light time at the flyby: 9 hours, at 33 AU from Earth.
- Encounter data was returned at 600 bps over about 9 months.

## The Deep Space Network (DSN)

The DSN is NASA's international array of giant radio antennas used to track, command and receive data from spacecraft at the Moon and beyond. It also supports radar and radio astronomy observations.

### From NASA's DSN page
- **Three complexes:** Goldstone (California), Canberra (Australia) and Madrid (Spain).
- **Management:** the Jet Propulsion Laboratory (JPL), run by Caltech, manages it for NASA's Space Communications and Navigation (SCaN) Program.
- **Navigation:** location and speed are inferred from changes in the spacecraft's radio signal frequency (the Doppler effect, like a passing siren). This is the radio-tracking half of New Horizons' navigation.
- **Antennas:** the largest are 70 m (230 ft) dishes at each complex. The NSSDCA notes that New Horizons data was received by a 70 m DSN dish.
- **Visitors and education:** each facility has a visitor center, and Goldstone's Apple Valley Radio Telescope is used for teaching.

### How the DSN works
*Verified against NASA's "What is the Deep Space Network?" and "How NASA's Deep Space Network Supports the Agency's Missions" pages.*

- **Three sites, about 120 degrees apart in longitude:** Goldstone (near Barstow, California), near Madrid (Spain) and near Canberra (Australia). Before a spacecraft sinks below the horizon at one site, another site picks up the signal, so coverage is continuous as Earth rotates.
- **Many antennas per site:** each complex has at least four antenna stations with large parabolic dishes and ultra-sensitive receivers that detect very faint signals. Each site has one 70 m (230 ft) antenna, the largest and most sensitive, able to track a spacecraft tens of billions of miles from Earth.
- **Shared and scheduled:** many missions use the network, so passes are negotiated between mission schedulers and DSN staff. Critical events (launches, landings, planetary maneuvers) need constant contact, and planning typically starts 12 to 15 weeks ahead. If demand overlaps, a spacecraft can store data onboard and send it later.
- **History:** the forerunner began in 1958 with portable tracking stations (California, Nigeria, Singapore) that received telemetry from Explorer 1. The DSN has run continuously since 1963.
- **Live view:** the DSN Now / SCaN Now web apps show which dish is talking to which spacecraft.

### What the DSN does (functions)
1. **Command (uplink):** send instructions to a spacecraft, such as new sequences or a clock update like the TKU.
2. **Telemetry and science (downlink):** receive images, instrument data and engineering data. New Horizons sent its Pluto encounter data at 600 bps over about 9 months.
3. **Tracking and navigation:** measure Doppler shift (speed) and ranging (distance) to fix a trajectory.
4. **Health and safety monitoring:** NASA describes the DSN as much more than big antennas: it is a system for commanding, tracking and monitoring spacecraft health.
5. **Radio science:** probe asteroids and the interiors of planets and moons using how a spacecraft's radio signal changes.
6. **Radar and radio astronomy:** observations of the solar system and the wider universe.

### Does the DSN have networking equipment in space?
**No. The DSN itself is entirely ground-based:** the antennas, receivers and operations systems are on Earth, and the spacecraft being tracked are its "customers". Related space-based networking does exist, but it is separate from the DSN dishes:

- **Spacecraft radios:** every mission carries its own transponder and antenna (for New Horizons, a DSN-compatible X-band link). This is the spacecraft end of the link, not DSN equipment.
- **Delay/Disruption Tolerant Networking (DTN):** NASA's store-and-forward protocol for the "Solar System Internet". Each network node stores data until the next node is available, like email waiting in an outbox. NASA says DTN became an operational service in both the Near Space Network and the DSN with the DTN Project's completion in January 2026. PACE (2024) was the first NASA Class-B mission to use DTN operationally for housekeeping telemetry.
- **Planned relays:** NASA plans a lunar network (LunaNet, with Near Space Network lunar relay services) and a DTN-enabled Mars relay orbiter, planned to be operational by 2030.
- **Near Space Network:** the DSN works in tandem with it. It is managed by NASA Goddard and serves missions closer to Earth.

*Source for this section: NASA's [Delay/Disruption Tolerant Networking](https://www.nasa.gov/communicating-with-missions/delay-disruption-tolerant-networking/) page. The statement that the DSN has no space-based hardware is an inference from NASA's description of the DSN as an array of antennas on three ground complexes.*

### Background knowledge (not verified against NASA pages)
- Doppler gives speed along the line of sight. Ranging (round-trip signal time) gives distance. Together they give a very precise trajectory.
- Signals from a spacecraft as far away as New Horizons are extremely weak, which is why large dishes are needed. The 600 bps rate from Pluto reflects this.
- Existing Mars orbiters act as relays for rovers and landers, passing their data on to Earth. This is the closest thing to networking hardware in space today, but those orbiters are spacecraft in their own right, not part of the DSN.
- At 4.5 hours one-way light time, a command sent to New Horizons at Pluto takes that long to arrive, so the spacecraft must run pre-loaded sequences on its own.

### Why was New Horizons' downlink only 600 bps at Pluto?
**From NASA's NSSDCA page (2006-001A):**
- At Pluto, data was sent over X-band at 600 bps through the 2.1 m high-gain antenna to a 70 m DSN dish.
- The flyby was 33 AU from Earth (9 hours round-trip light time), and encounter data took about 9 months to return.
- Power was limited: the RTG provided about 228 W at encounter, and the instruments were duty-cycled because of it.

**Why that forces a low rate (general physics, not from NASA's pages):**
1. **Distance:** radio power spreads out and falls with the square of the distance. At 33 AU the signal is about 33² ≈ 1,100 times weaker than at 1 AU.
2. **Small transmitter and dish:** the spacecraft can only send a few watts through a 2.1 m dish. (I recall about 12 W, which is unverified.)
3. **Noise:** each bit needs enough energy compared with the background noise to be decoded reliably. When the signal is faint, the only way to keep enough energy per bit is to send fewer bits per second, so each bit lasts longer.
4. **Receiver size:** a bigger receiving dish collects more of the faint signal, which is why a 70 m DSN dish was used.

The 600 bps was the rate chosen for the Pluto encounter, not a fixed limit. The rate that can be sustained drops as the spacecraft gets farther away. At 600 bps, a 1 MB image takes about 3.7 hours.

## LORRI: the Long Range Reconnaissance Imager

*Sources: APL's [New Horizons payload page](https://pluto.jhuapl.edu/Mission/Spacecraft/Payload.php) (APL), Wikipedia's [LORRI](https://en.wikipedia.org/wiki/Long_Range_Reconnaissance_Imager) article for the optical details (Wikipedia), and the calibration paper [Weaver et al. 2020, In-Flight Performance and Calibration of LORRI](https://arxiv.org/abs/2001.03524) (Weaver 2020).*

### The instrument
- **Telescope and camera (APL):** a panchromatic (black-and-white) camera with a 20.8 cm (8.2 in) aperture telescope focusing visible light onto a CCD (charge-coupled device). Mass 8.8 kg, average power 5.8 W. APL calls it essentially a digital camera with a large telephoto telescope, built for the cold of the Kuiper Belt.
- **No filters, no moving parts (APL):** operators aim LORRI by pointing the whole spacecraft at the target. This is why flyby timing and pointing mattered so much (see the TKU section).
- **Silicon carbide construction (APL):** keeps the mirrors in focus through the extreme temperature drops on the way to and beyond Pluto.
- **Optics (Wikipedia):** Ritchey-Chretien design, 1024 x 1024 pixel CCD, field of view about 0.29 degrees, about 5 microradians per pixel.
- **What a CCD is (general background, not from the sources above):** a silicon chip that turns light into a digital image. It is a grid of tiny light-sensitive pixels, and each pixel collects electrons freed by the photons that hit it, so brighter light gives more charge. After the exposure the chip shifts each pixel's stored charge across the grid to one readout point (the "charge-coupled" part), and a converter turns each charge into a number. LORRI records 12 bits per pixel (Wikipedia).
- **Why a CCD suits LORRI:** it can collect faint light over a long exposure, which matters in the dim sunlight at Pluto. Wikipedia lists LORRI's sensor as a back-illuminated CCD (a design that gathers light more efficiently) and says it is chilled by a passive radiator on the side of the spacecraft facing away from the Sun. Its pixels sit at fixed, known positions, which lets navigators measure where Pluto appears against the stars very precisely.
- **No color:** LORRI has no color filters, so it records only brightness. This trades color for sensitivity. Ralph's MVIC camera provides the color images.

### What it did at Pluto (APL)
- **About 180 days before closest approach:** LORRI took New Horizons' first images of the Pluto system. Pluto and its moons were little more than bright dots, but these system-wide views helped navigators keep the spacecraft on course and helped scientists refine the orbits of Pluto and its moons.
- **About 60 days before:** around mid-May 2015, LORRI images began to surpass Hubble-quality resolution.
- **At closest approach:** LORRI imaged selected sunlit areas at about 50 m resolution (football-field size).
- **Science:** crater counts and surface history, a search for geysers and activity, and hazes in Pluto's atmosphere.

### Its role in navigation
A far-away target is only a dot, but a dot against known stars still shows where Pluto is, as seen from the spacecraft. That direction is measured much more precisely than Earth-based observations of Pluto's position. Wikipedia says LORRI is used for navigation, especially to refine the location of a flyby target, and that its data were used the same way for the 2019 Arrokoth flyby.

### How LORRI navigation works when everything is moving
*General background on optical navigation, not verified against NASA or APL pages, except where marked.*

LORRI does not track motion itself. It takes still images, and the navigation team measures where Pluto sits against the background stars.
1. **Stars as a fixed backdrop:** stars are so far away that over weeks or months they stay put at LORRI's precision. Their positions come from star catalogs, so every frame has reference points.
2. **Pluto moves against the stars:** Pluto and its moons orbit the Sun, and the spacecraft is moving too, so Pluto shifts against the stars from one image to the next. That shift is the signal.
3. **Time-stamped images:** each image has a known time, so measuring Pluto's position among the stars gives its direction from the spacecraft at that time.
4. **Knowing where LORRI points:** LORRI is fixed to the spacecraft body (APL), so the spacecraft's orientation must be known. The NSSDCA says star cameras are mounted on the spacecraft for navigation, and the stars in LORRI's own frame also help pin down the pointing.
5. **Combining with radio tracking:** the directions go into the same orbit solution as the DSN Doppler and ranging data. As the spacecraft moves, the direction to Pluto changes, which gives a precise position. APL says the early images helped navigators keep the spacecraft on course and refine the orbits of Pluto and its moons.

**Why the precision is high:** LORRI's pixels are about 5 microradians wide (Wikipedia). About 180 days before closest approach (APL), the spacecraft was roughly 2 x 10^8 km from Pluto (my estimate, assuming about 13 km/s), so one pixel spans roughly 1,000 km. Finding the center of a bright dot to a fraction of a pixel narrows the position down much further.

### Which reference stars does LORRI use?
There is no single reference star system. Reference stars are used in two different ways.

**1. Calibrating the camera (Weaver 2020):**
- The team images star clusters whose stars are listed in astrometric surveys (catalogs of precise star positions). This measures sharpness (the point-spread function), sensitivity and geometric distortion across the whole field, and repeats every year of the mission.
- **Messier 7** (also called NGC 6475, the Ptolemy cluster) was used from 2006 to 2013. In 2013 the team switched to **NGC 3532** (the Wishing Well cluster) because it has more stars, giving better coverage across the CCD. Both were observed in 2013 so the later results could be compared with the earlier ones.
- Results: LORRI's distortion stays at 0.3% or less across the field, and the paper reports no detectable change in sensitivity or optical performance (at the 1% level) since launch.
- These clusters test the camera. They are not the stars used to navigate to Pluto.

**2. Navigating to a target:**
- Navigators measure Pluto against whatever catalogued stars appear in the same frames. Weaver 2020 describes LORRI as the mission's primary optical navigation camera.
- Wikipedia says New Horizons was heading toward the constellation Sagittarius and that Arrokoth was seen against a crowded star background in that direction. That suggests plenty of reference stars near the targets, but I have not confirmed this for Pluto.
- Weaver 2020 also describes a template of the star field around Arrokoth, made by combining 96 images taken in 2017. Subtracting it from late-2018 images removed the nearby stars so Arrokoth could be detected for optical navigation.
- I did not confirm which star catalog the navigation team used.

### How LORRI avoids motion blur (Weaver 2020)
- **No reaction wheels:** because of power limits, New Horizons holds its pointing by firing hydrazine thrusters to stay inside a small allowed wobble (a "deadband"), with firings about every 3 s or 6 s depending on the deadband.
- **Short exposures:** typical science exposures are 0.10 to 0.15 s, chosen to keep pointing smear to one pixel or less.
- **Scans near closest approach:** when the spacecraft swept Ralph across Pluto, LORRI took images every 1 or 3 seconds with even shorter exposures. The overlapping images were added together, and deconvolution (a sharpening calculation) recovered resolution. One of these scans gave the highest-resolution imaging of Pluto during the flyby.
- **Long exposures:** a special "relative control mode", developed shortly after launch, fires the thrusters about 80 times a minute to hold LORRI on a fixed spot of sky, allowing exposures up to about 65 s. Combined with 4 x 4 pixel binning and about 100 added images, LORRI can detect faint targets down to about magnitude V = 22.
- **Pixel scale:** the paper gives 1.02 arcsec per pixel, matching the roughly 5 microradians quoted above.
