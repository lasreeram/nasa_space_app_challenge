## Mission Overview

New Horizons was the first spacecraft to explore Pluto and its moons up close (flyby on July 14, 2015). In early 2019 it flew past Arrokoth (2014 MU69), the most distant object ever explored up close. It was the first mission in NASA's New Frontiers program (which also includes Juno and OSIRIS-REx).

| Item | Detail |
|---|---|
| Objectives | Pluto flyby, Kuiper Belt Object flyby |
| Management | NASA / Johns Hopkins University Applied Physics Laboratory (APL) |
| Spacecraft mass | 1,054 lb (478 kg) |
| Power | RTG, ~250 W at launch, ~200 W at Pluto (a Cassini spare) |
| Launch vehicle | Atlas V 551 (AV-010), Cape Canaveral LC-41 |
| Launch | Jan. 19, 2006, 19:00:00 UT |
| Launch velocity | ~36,400 mph (58,536 km/h), highest launch speed of any human-made object relative to Earth |

### Key dates
- **Jan. 19, 2006**: Launch
- **Feb. 28, 2007**: Jupiter gravity assist (+~9,000 mph, shortened the trip by ~3 years)
- **July 14, 2015**: Pluto flyby
- **Jan. 1, 2019**: Arrokoth flyby
- **Oct. 1, 2024**: 60x Earth's distance from the Sun (twice as far as Pluto at flyby)

### Scientific instruments
**Remote-sensing instruments**
- **Ralph**: main color imager and infrared mapping spectrometer. Maps surface composition (nitrogen, methane and water ices) and takes color images. Two parts: MVIC (visible color camera) and LEISA (near-infrared spectral imager).
- **Alice**: ultraviolet imaging spectrometer. Studies the composition and structure of Pluto's atmosphere, looks for escaping gas, and searches for atmospheres around other objects.
- **LORRI** (Long-Range Reconnaissance Imager): telescopic black-and-white camera with a ~20 cm mirror. Takes the highest-resolution images, spots targets from far away, and helps with navigation.

**Radio science**
- **REX** (Radio Science Experiment): uses the spacecraft's radio link to Earth instead of a separate sensor. Measures atmospheric temperature and pressure and surface temperature, and helps determine mass.

**Particle instruments**
- **SWAP** (Solar Wind Around Pluto): measures how the solar wind interacts with Pluto's atmosphere, including how fast the atmosphere escapes.
- **PEPSSI** (Pluto Energetic Particle Spectrometer Science Investigation): detects energetic ions, especially from Pluto's atmosphere, and measures their energy and composition.

**Dust**
- **SDC** (Student Dust Counter): student-built and student-operated dust detector. Counts dust grains along the whole trajectory, mapping dust density from the inner solar system into the Kuiper Belt.

## Flyby Timing: the Timing Knowledge Update (TKU) and Navigation

*Source note: the TKU description comes from the researcher's own notes. The navigation details are background knowledge and not verified against NASA pages. The NSSDCA confirms only the DSN X-band link and the onboard star cameras.*

### The problem
New Horizons flew by Pluto on a pre-loaded command sequence (the "core load") timed to its onboard clock. The cameras and spectrometers are fixed to the spacecraft body, so they point wherever the whole spacecraft points. To center Pluto and its moons in the instrument boresight, the spacecraft had to reach closest approach within +/-540 s (+/-9 min) of the planned time. Radio signals took about 4.5 hours each way, so live control was impossible.

### The Timing Knowledge Update
- The TKU adjusts the onboard clock so the spacecraft believes it is slightly ahead of or behind its real position in the core load.
- All planned flyby activities slide earlier or later together, by up to 540 s, to line up with the final, best predicted arrival time.
- It changes one number instead of rewriting a tested sequence, so activities keep the same order and spacing.

### Navigation data
1. **Radio tracking:** Deep Space Network antennas measure Doppler shift (speed along the line of sight) and ranging (distance), fixing the trajectory relative to Earth.
2. **Optical navigation:** in the months before the encounter, the spacecraft imaged Pluto and its moons against known star fields, giving the direction to Pluto with more precision than Earth-based tracking.
3. **Combination:** the navigation team combined both into a best predicted arrival time, which set the TKU and any course corrections.

```mermaid
flowchart LR
    A[DSN Doppler and ranging] --> C[Orbit determination]
    B[Images of Pluto against star background] --> C
    C --> D[Best predicted arrival time]
    D --> E[TKU: shift onboard clock up to 540 s]
    C --> F[Course corrections, if needed]
    E --> G[Sequence aligned with true arrival: Pluto and moons in boresight]
```

### Facts from NASA's NSSDCA (2006-001A)
- Closest approach: 11:48:45 UT on July 14, 2015, at 12,500 km from Pluto and 11 km/s relative velocity (not 14 km/s, which is the speed relative to the Sun).
- Closest approach to Charon: about 27,000 km.
- Round-trip light time at the flyby: 9 hours, at 33 AU from Earth.
- Encounter data was returned at 600 bps over about 9 months.
- Source: [NSSDCA New Horizons page](https://nssdc.gsfc.nasa.gov/nmc/spacecraft/display.action?id=2006-001A)

## The Deep Space Network (DSN)

The DSN is NASA's international array of giant radio antennas used to track, command and receive data from spacecraft at the Moon and beyond. It also supports radar and radio astronomy observations.

### From NASA's DSN page
- **Three complexes:** Goldstone (California), Canberra (Australia) and Madrid (Spain).
- **Management:** the Jet Propulsion Laboratory (JPL), run by Caltech, manages it for NASA's Space Communications and Navigation (SCaN) Program.
- **Navigation:** location and speed are inferred from changes in the spacecraft's radio signal frequency (the Doppler effect, like a passing siren). This is the radio-tracking half of New Horizons' navigation.
- **Antennas:** the largest are 70 m (230 ft) dishes at each complex. The NSSDCA notes that New Horizons data was received by a 70 m DSN dish.
- **Visitors and education:** each facility has a visitor center, and Goldstone's Apple Valley Radio Telescope is used for teaching.

### Background knowledge (not verified against NASA pages)
- The three sites are about 120 degrees apart around the globe, so at least one antenna can see a distant spacecraft as Earth rotates.
- Doppler gives speed along the line of sight. Ranging (round-trip signal time) gives distance. Together they give a very precise trajectory.
- Signals from a spacecraft as far away as New Horizons are extremely weak, which is why large dishes are needed. The 600 bps rate from Pluto reflects this.
- Many missions share the DSN, so contact passes are scheduled in advance.

Source: [NASA Deep Space Network](https://www.nasa.gov/communicating-with-missions/dsn/)
