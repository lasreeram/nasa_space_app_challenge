# Doppler Effect: Tracking a Spacecraft by Its Radio Signal

The Doppler effect is the change in a wave's observed frequency when the source and observer move toward or away from each other. A siren sounds higher as it approaches and lower as it moves away. Radio waves behave the same way.

For a spacecraft with line-of-sight speed *v*, the shift is approximately **Δf = f × v / c**.

- Approaching Earth: signal arrives at a **higher** frequency.
- Receding from Earth: signal arrives at a **lower** frequency.
- Only the speed along the line of sight is measured. Sideways motion is not seen directly.

## How the Deep Space Network uses it

- With a stable 8.4 GHz signal, a speed change of 1 m/s shifts the frequency by about 28 Hz, which is easy to measure.
- In two-way tracking the signal is sent up and returned, so the shift is roughly doubled.
- Over days and weeks, the changing Doppler pattern reveals the direction of travel. Optical images against the stars add the sideways direction to Pluto.
- Doppler and ranging feed the predicted arrival time that sets the New Horizons Timing Knowledge Update.
